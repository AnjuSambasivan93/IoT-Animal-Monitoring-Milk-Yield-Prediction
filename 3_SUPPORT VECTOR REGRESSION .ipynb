```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
dataset=pd.read_csv('editedab.csv',index_col=0)
```

```python
dataset.head()
```

```python
dataset.describe()
```

```python
S = dataset.iloc[:,1:5].values
t = dataset.iloc[:,len( dataset.iloc[0])-1].values
```

```python
print(S)
print('\n')
print(t)
```

```python
from sklearn.model_selection import train_test_split  
X_train, X_test, y_train, y_test = train_test_split(S, t, test_size=0.2, random_state=0)
```

```python
x1=X_train.shape
print(x1)
x2=X_test.shape
print(x2)
x3=y_train.shape
print(x3)
x4=y_test.shape
print(x4)
```

```python
print(X_train)
print('\n')
print(X_test)
```

```python
# The implementation is based on libsvm.

# 1) C:- Penalty parameter C of the error term

# 2) cache_size:- Specify the size of the kernel cache (in MB).
#   We can use available memory (called kernel cache) to store some recently used.
#   Then, some kernel elements may not need to be recalculated.

# 3) class_weight : Set the parameter C of class i to class_weight[i]*C for SVC. 
#    If not given, all classes are supposed to have weight one. 

# 4) coef0 : float, optional (default=0.0). 
#    Independent term in kernel function. It is only significant in ‘poly’ and ‘sigmoid’.

# 5) decision_function:- "Distance of the samples X to the separating hyperplane."
#  return a one-vs-rest (‘ovr’) decision function of shape
#  (n_samples, n_classes) as all other classifiers, 
#  The desion function tells us on which side of the hyperplane
#  generated by the classifier we are
# (and how far we are away from it).
# Based on that information, the estimator then label the examples with the corresponding label.

# 6) degree : int, optional (default=3)
#    Degree of the polynomial kernel function (‘poly’). Ignored by all other kernels

# 7) gamma : float, optional (default=’auto’) Kernel coefficient for ‘rbf’, ‘poly’ and ‘sigmoid’.
#  Current default is ‘auto’ which uses 1 / n_features.

#  8) auto_deprecated’, a deprecated (Disapproval) version of ‘auto’ is used as a default 
#    indicating that no explicit value of gamma was passed.

# 9) kernel : string, optional (default=’rbf’)(Radial basis function)
#   Specifies the kernel type to be used in the algorithm. 
#   It must be one of ‘linear’, ‘poly’, ‘rbf’, ‘sigmoid’. 

# 10) max_iter : int, optional (default=-1)
#    Hard limit on iterations within solver, or -1 for no limit.

# 11) probability : boolean, optional (default=False)
#   This must be enabled prior to calling fit, and will slow down that method.
#   (confidence in the label) 
#   If this parameter is set to True, 
#  libsvm will train a probability transformation model on top of the SVM's outputs.

# 12) random_state : int, RandomState instance or None, optional (default=None)
#  used when shuffling the data for probability estimates(when probability is set to True)

# 13) verbose : bool, default: False Enable verbose output.
#  used to report more information about an operation in your program.
# Verbose is a general programming term for produce lots of logging output. 
# meaning is "tell me everything about what you are doing all the time". 

# 14) shrinking : boolean, optional (default=True) Whether to use the shrinking heuristic. 
# To save the training time, the shrinking technique
# tries to identify and remove some bounded elements, 
# so a smaller optimization problem is solved

# 15) tol : float, optional (default=1e-3) Tolerance for stopping criterion.


from sklearn import svm
support=svm.SVR(C=1,kernel='linear',epsilon=1)
support.fit(X_train, y_train)
```

```python
y_pred = support.predict(X_test)  
```

```python
# COMPLEXITY
# Support Vector Machines are powerful tools, but their compute and storage requirements increase rapidly with 
# the number of training vectors.
# The core of an SVM is a quadratic programming problem (QP), separating support vectors from the rest of the training data.
# The QP solver used by this libsvm-based implementation scales between O(n_features* n^2_samples) and O(n_features* n^3_samples)
# If the data is very sparse n_features should be replaced by the average number of non-zero features in a sample vector
```

```python
diff=y_test-y_pred
print(diff)
plt.plot(diff)
plt.show()
```
